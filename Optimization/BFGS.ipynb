BFGS is perhaps the single most popular 2nd-order differentiation optimization algorithm.

The main advantage of Newton/2nd-order differentiation methods is that whilst Gradient Descent or Stochastic Gradient Descent takes hundred of thousands or sometimes millions of steps to converge, Newton or Quasi-Newton methods converge in fractions of those counts. 
The main problem, however, with such method is that computation required is HUGE in comparision of gradient methods which require very less computation power.


For more detailed breakdown on these methods:

Newton Methods of Optimization: https://en.wikipedia.org/wiki/Newton%27s_method_in_optimization
Quasi Newton methods of Optimization: https://towardsdatascience.com/bfgs-in-a-nutshell-an-introduction-to-quasi-newton-methods-21b0e13ee504/

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

```python
def func(x, y):
    fun = 100*(x**2) + (y)**(-2)
    return fun
```

```python
x = sp.Symbol('x')
y = sp.Symbol('y')
def gradient(x, y):
    gradient = sp.Matrix([
        sp.diff(func(x, y), x),
        sp.diff(func(x, y), y)
    ])
    return gradient
gradient = np.array(gradient(x, y))
print(gradient)
```

```python
Hessian = sp.hessian(func(x, y), (x, y))
Hessian = np.array(Hessian)
print(Hessian)
```

Unlike in Gradient Methods, where we need to define the step size as a hyperparameter in advance, BFGS needs line search method for finding the appropriate step size $\alpha$

```python
def line_search (fun, x ,y, p, grad):
    a = 1
    c1 = 1e-6
    c2 = 0.99
    xn = x + a*p
    yn = y+ a*p
    grad_0 = grad(x, y)
    grad_n = grad(xn, yn)
    while fun(xn, yn) >= fun(x,y) + (c1*a*grad_0.T@p) or grad_n.T@p <= c2*grad_0.T@p:
        a *= 0.5
        xn = x + a*p
        yn = y+ a*p
        grad_n = grad(xn, yn)
    return a
```
