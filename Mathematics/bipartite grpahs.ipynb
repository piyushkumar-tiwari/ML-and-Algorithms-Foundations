Bipartite graphs are special types of graphs $G(E, V)$ in which graph can be $\textbf{disjointed}$ into 2 subsets of vertices $V_i \text{ and } V_j$ such that all the edges $E$ goes from $V_i \text{ to } V_j$ and vice versa but not from set  $V_i \text{ to } V_i \text{ or } V_j \text{ to } V_j$.

There is another way to look them is through a property called $\textbf{chromatic number}$.
For a graph $G(E, V)$, chromatic number $\mathcal{X}(G)$ is the minimal number of colors required to label all the vertices in the graph such that no two adjacent vertices have the same color. For a bipartite graph:
$\mathcal{X}(G) \leq 2 $
